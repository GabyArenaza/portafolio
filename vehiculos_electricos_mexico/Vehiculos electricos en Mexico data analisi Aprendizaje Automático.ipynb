### 7 Aprendizaje Automático

 La Regresión Lineal simple se va a encargar de encontrar la relación entre xs y ys

```python
# Se importan las librerías de Python
import numpy  as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
ventas_anuales['year'] = ventas_anuales.index

datos = ventas_anuales.loc[
    ventas_anuales.index.isin(range(2016, 2026))
]

xs = datos.index.to_numpy()
ys = datos['PHEV'].to_numpy()


print(f'xs ==>', xs)
print(f'ys ==>' ,ys)

print(len(xs))
print(len(ys))
```

```python
# Operaciones para una Regresión Lineal Simple
# conseguir  el Beta y el alpha
# con polyfit que usa mínimos cuadrados
regresion_lineal = np.polyfit(xs,ys,1)  # polinomio grado 1
print(f"Regresion Lineal =",regresion_lineal)

```

```python
# Asignar a Beta y alpha los valores de la regresión lineal
beta,alpha =  regresion_lineal
print(f"Beta (pendiente) = {beta:.2f}")
print(f"Alpha (intercepto) = {alpha:.2f}")
```

```python
# Graficamos valores y regresión lineal simple
plt.figure(figsize=(8,4))
plt.plot(xs, ys, 'o', color="green", label= "Datos")
plt.plot(xs, beta*xs+alpha, color="blue",label="Regresión Lineal")
plt.axhline(y=0, color="red")
plt.axvline(x=2016, color="red")
plt.title("Grafica de regresión lineal simple para ventas de vehículos hibridos en México (2016 -2025)")
plt.xlabel("xs = año")
plt.ylabel("ys = ventas de PHEV")
plt.legend()
plt.grid()
plt.show()
```

```python
# Se genera la RED NEURONAL
# se crea el modelo
modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

modelo.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

```python
# Comenzando el entrenamiento
historial = modelo.fit(
    xs_norm,
    ys_norm,
    epochs=500,
    verbose=False
)
```
